```python
import pandas as pd
from datetime import date

#read collected data 
data = pd.read_csv("data.csv", header= None)

#convert each row of data into a matrix
matrix = []
for i in range(len(data)):
    list = data.iloc[i - 1].to_list()
    list_length = len(list)
    matrix.append(list)

matrix
```

```python
#generate the calender list from November 1 2021 to October 1 2026
start = date(2021, 11, 1)
end = date(2026, 10, 1)

date_list = []
current = start

while current <= end:
    date_list.append(current)
    if current.month == 12:
        current = date(current.year + 1, 1, 1)
    else:
        current = date(current.year, current.month + 1, 1)

date_list
```

```python
df = pd.DataFrame(matrix).T
df.index = date_list
df = df.rename(columns= {0: 'bsp_interest_rate', 
                         1: 'cpi_headline',
                         2: 'food_cpi',
                         3: 'transport_cpi',
                         4: 'inflation_rate_yoy',
                         5: 'php_usd_exchange',
                         6: 'global_oil_price'})

df
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

corr_matrix = df.corr(numeric_only=True)
spearman = df.corr(method='spearman', numeric_only=True)

plt.figure(figsize=(8, 6))
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    vmin=-1,
    vmax=1
)

plt.title('Correlation Matrix Heatmap')
plt.show()
```
