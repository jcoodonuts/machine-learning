```python
import pandas as pd
from datetime import date

start = date(2021, 11, 1)
end = date(2026, 10, 1)

date_list = []
current = start
```

```python
#generates the calender list from November 1 2021 to October 1 2026
while current <= end:
    date_list.append(current)
    if current.month == 12:
        current = date(current.year + 1, 1, 1)
    else:
        current = date(current.year, current.month + 1, 1)

date_list
```
